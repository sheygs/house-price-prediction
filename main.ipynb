```python
import pandas as pd

# save the file path
file_path = './datasets/melb_data.csv'

# read & store the data in a DataFrame
melbourne_data = pd.read_csv(file_path)

# print a summary
melbourne_data.describe()
```

```python
# display column names; might have some missing values
# (some houses for which some variables weren't recorded)
print(f'columns: {melbourne_data.columns}')
# print("before cleaning: ", melbourne_data.count(axis=0))

# removes rows with missing values (NaNs) from a DataFrame
# axis=0 -> drop rows; axis=1 -> drop columns

# `dropna` drops missing/"not available" values
melbourne_data = melbourne_data.dropna(axis=0)

#print("after cleaning: ", melbourne_data.count(axis=0))
```

```python
# select the 'prediction' target
y = melbourne_data.Price

# features - columns inputted into our model and later used to make predictions
# In our case, those would be the columns used to determine the home price
# Sometimes, you will use all columns EXCEPT the target as features
melbourne_features = ['Rooms', 'Bathroom', 'Landsize', 'Lattitude', 'Longtitude']
```

```python
# By convention, this data is called `X`
X = melbourne_data[melbourne_features]
X.describe()
```

```python
# display entries
print(f"Entries are: \n{X}")
```

```python
from sklearn.tree import DecisionTreeRegressor

# Model definition - specify a number for `random_state` to ensure same results each run.
# You may use any number, and model quality won't depend meaningfully on exactly what value you choose.
melbourne_model = DecisionTreeRegressor(random_state=1)

# Fit the model
melbourne_model.fit(X, y)
```

```python
print("Making predictions for the following 5 houses\n")
# display first 5 entries
# print(X.head())

# Compare the top few predictions to the actual home values (in y) for those same homes. Anything surprising?
print(
    f"\nFirst in-sample home prices prediction are: {melbourne_model.predict(X.head())}"
)

print("Actual target values for those homes:", y.head().tolist())
```

```python
# Model Validation - measuring the performance of your model, so you can test and compare alternatives

# Bad Approach - In-sample approach using a single sample to build/train & evaluate the model
from sklearn.metrics import mean_absolute_error
predicted_home_prices = melbourne_model.predict(X)
mean_absolute_error(y, predicted_home_prices)
```

```python
# The Good/Right Approach
from sklearn.model_selection import train_test_split

# split data into training and validation data, for both features and target
# The split is based on a random number generator. Supplying a numeric value to
# the random_state argument guarantees we get the same split every time we
# run this script

# print(train_test_split(X, y, random_state=0))
train_X, val_X, train_y, val_y = train_test_split(X, y, random_state=0)

print("total count X: ", len(X))
print("total count y: ", len(y))
print("train X count: ", len(train_X))
print("test X count: ", len(val_X))
print("train y count: ", len(train_y))
print("test y count: ", len(val_y))

# model definition
model_def = DecisionTreeRegressor(random_state=1)

# fit model
model_def.fit(train_X, train_y)

# display pridicted price on validation data
predicted_prices = model_def.predict(val_X)
print(f"mean absolute error: {mean_absolute_error(val_y, predicted_prices)}")
```

```python
# Overfitting and Underfitting
# Fine-tuning your model for better performance.
def get_mae(max_leaf_nodes, train_X, val_X, train_y, val_y):
    model = DecisionTreeRegressor(max_leaf_nodes=max_leaf_nodes, random_state=1)
    model.fit(train_X, train_y)
    predicted_value = model.predict(val_X)
    mae = mean_absolute_error(val_y, predicted_value)
    return (mae)
```

```python
# compare MAE with differing values of max_leaf_nodes
candidate_max_leaf_nodes = [5, 50, 500, 5000]

for max_leaf_node in candidate_max_leaf_nodes:
    val = get_mae(max_leaf_node, train_X, val_X, train_y, val_y)
    print(f"Max leaf nodes: {max_leaf_node}  \t\t Mean Absolute Error:  {val:.0f}")

# shorter/cleaner approach
scores = {
   leaf_size: get_mae(leaf_size, train_X, val_X, train_y, val_y) for leaf_size in candidate_max_leaf_nodes
}

print(f'scores: {scores}')

best_tree_size = min(scores, key=scores.get)

# Of the options listed, 500 is the optimal number of leaves.
print(f'best tree size: {best_tree_size}')
```

```python
# final model
final_model = DecisionTreeRegressor(max_leaf_nodes=best_tree_size, random_state=1)
final_model.fit(X, y)
```

```python
# Optimisation - using a Random Forest
from sklearn.ensemble import RandomForestRegressor

forest_model = RandomForestRegressor(random_state=1)
forest_model.fit(train_X, train_y)
melb_preds = forest_model.predict(val_X)
print(mean_absolute_error(val_y, melb_preds))
```
