```python
import pandas as pd
```

```python
"""
Creating data
There are two core objects in pandas: the DATAFRAME and the SERIES.

DataFrame
A DataFrame is a table. It contains an array of individual entries, each of
which has a certain value. Each entry corresponds to a row (or record) and a column.
"""
```

```python
# For example, consider a simple DataFrame

pd.DataFrame({'Yes': [12, 24], 'No': [13, 26]})

# In this example, the "0, No" entry has the value of 13. The "0, Yes" entry has a value of 12, and so on.
```

```python
# Example of DataFrame whose values are strings
pd.DataFrame({'Sheygs' : ['I loved it', 'It was sweet'], 'John': ['Pretty good', 'Bland']})
```

```python
"""
The dictionary-list constructor assigns values to the column labels,
but just uses an ascending count from 0 (0, 1, 2, 3, ...) for the row labels.
Sometimes this is OK, but oftentimes we will want to assign these labels ourselves.
The list of row labels used in a DataFrame is known as an INDEX.
We can assign values to it by using an `index` parameter in our constructor:
"""
```

```python
pd.DataFrame({'Sheygs': ['I loved it', 'It was sweet'], 'John': ['Pretty good', 'Bland']}, index=['Product A', 'Product B'])
```

```python
"""
Series: It is a sequence of data values.
If DataFrame is a table, a Series is a list

"""
```

```python
pd.Series([20, 25, 30, 40])
```

```python
"""
A Series is, in essence, a single column of a DataFrame.
So you can assign row labels to the Series the same way as before,
using an `index` parameter. However, a Series does not have a
column name, it only has one overall `name`:
"""
```

```python
pd.Series([20, 25, 30, 40], index=['2014 Sales', '2015 Sales', '2016 Sales', '2017 Sales'], name='Product A')
```
