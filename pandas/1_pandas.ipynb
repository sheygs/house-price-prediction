### Creating data

There are two core objects in pandas: the **DataFrame** and the **Series**.

#### DataFrame

A DataFrame is a table. It contains an array of individual entries, each of
which has a certain value. Each entry corresponds to a row (or record) and a column.

```python
pd.DataFrame({'Yes': [12, 24], 'No': [13, 26]})
```

In this example, the "0, No" entry has the value of 13. The "0, Yes" entry has a value of 12, and so on.  DataFrame entries are not limited to integers. For instance, here's a DataFrame whose values are strings:

```python
pd.DataFrame({'Sheygs' : ['I loved it', 'It was sweet'], 'John': ['Pretty good', 'Bland']})
```

The dictionary-list constructor assigns values to the column labels,
but just uses an ascending count from 0 (0, 1, 2, 3, ...) for the row labels.
Sometimes this is OK, but oftentimes we will want to assign these labels ourselves.
The list of row labels used in a DataFrame is known as an **index**.
We can assign values to it by using an `index` parameter in our constructor:

```python
pd.DataFrame({'Sheygs': ['I loved it', 'It was sweet'], 'John': ['Pretty good', 'Bland']}, index=['Product A', 'Product B'])
```

#### Series

It is a sequence of data values. If DataFrame is a table, a Series is a list

```python
pd.Series([20, 25, 30, 40])
```

A Series is, in essence, a single column of a DataFrame.
So you can assign row labels to the Series the same way as before,
using an `index` parameter. However, a Series does not have a
column name, it only has one overall `name`:

```python
pd.Series([20, 25, 30, 40], index=['2014 Sales', '2015 Sales', '2016 Sales', '2017 Sales'], name='Product A')
```

**Note**: A dataFrame is actually just a bunch of "Series" glued together.

### Reading Data Files

```python
reviews = pd.read_csv('../pandas/datasets/winemag-data-130k-v2.csv')
```

We can use the `shape` attribute to check how large the resulting DataFrame is:

```python
# About 130,000 records split across 14 columns
reviews.shape
```

We can examine the contents of the resultant DataFrame using the `head()` command, which grabs the first five rows:

```python
reviews.head()
```

The `pd.read_csv()` function is well-endowed, with over 30 optional parameters you can specify.
For example, you can see in this dataset that the CSV file has a built-in index,
which pandas did not pick up on automatically. To make pandas use that column for the index
(instead of creating a new one from scratch), we can specify an `index_col`.

```python
reviews = pd.read_csv('../pandas/datasets/winemag-data-130k-v2.csv', index_col=0)

reviews.head()
```
