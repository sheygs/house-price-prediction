```python
# Import helpful libraries
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
# from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split

# Load the data
data_path = './datasets/train.csv'
data = pd.read_csv(data_path)

# separate the prediction target
y = data.SalePrice

# Create X (After completing the exercise, you can return to modify this line!)
features = ['LotArea', 'YearBuilt', '1stFlrSF', '2ndFlrSF', 'FullBath', 'BedroomAbvGr', 'TotRmsAbvGrd']

# Select columns corresponding to features, and preview the data
X = data[features]
X.head()

# Split into validation and training data
train_X, val_X, train_y, val_y = train_test_split(X, y, random_state=1)

# Old/Sub-Optimal Approach
# Specify Model
# dtr_model = DecisionTreeRegressor(random_state=1)
# Fit Model
# dtr_model.fit(train_X, train_y)

# Make validation predictions and calculate mean absolute error
# predicted_X_val = dtr_model.predict(val_X)
# dtr_val_mae = mean_absolute_error(predicted_X_val, val_y)
# print("Validation MAE when not specifying max_leaf_nodes: {:,.0f}".format(dtr_val_mae))

# Using best value for max_leaf_nodes
# dtr_model = DecisionTreeRegressor(max_leaf_nodes=100, random_state=1)
# dtr_model.fit(train_X, train_y)
# predicted_X_val = dtr_model.predict(val_X)
# dtr_val_mae = mean_absolute_error(predicted_X_val, val_y)
# print("Validation MAE for best value of max_leaf_nodes: {:,.0f}".format(dtr_val_mae))


# Define a random forest model
rf_model = RandomForestRegressor(random_state=1)
rf_model.fit(train_X, train_y)
rf_val_predictions = rf_model.predict(val_X)
rf_val_mae = mean_absolute_error(rf_val_predictions, val_y)

print("Validation MAE for Random Forest Model: {:,.0f}".format(rf_val_mae))
```

```python
# To improve accuracy, create a new Random Forest model which you will train on all training data
rf_model_full_data = RandomForestRegressor(random_state=1)

# fit the model on all data from the training data
rf_model_full_data.fit(X, y)

# read the "test" data
test_data_path = './datasets/test.csv'
test_data = pd.read_csv(test_data_path)
# test_data.describe()

# print(test_data.columns)

# create test_X which comes from test_data but includes only
# the columns you used for prediction. The list of columns
# is stored in a variable called 'features'
test_X = test_data[features]

# make predictions
predicted_test_values = rf_model_full_data.predict(test_X)

print(f'Predicted Test values: {predicted_test_values}')

# save predictions in `csv` format
output = pd.DataFrame({ 'Id': test_data.Id, 'SalePrice': predicted_test_values })
output.to_csv('output.csv', index=False)
```
